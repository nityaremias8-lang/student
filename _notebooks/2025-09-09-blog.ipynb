    <h2>📅 First Day Recap</h2>
    <ul>
      <li>⬇️ First, I downloaded <strong>VS Code</strong> and practiced opening the built-in <code>terminal</code>. At first it felt intimidating — a blank black screen blinking at me 😅 — but soon I realized it’s just a way to talk to my computer more directly.</li>
      <li>⌨️ I learned the “survival commands” for navigating the file system: <code>mkdir</code>, <code>cd</code>, <code>pwd</code>, and <code>ls</code>. These gave me the power to move around and create folders without relying on the mouse 🖱️.</li>
      <li>🐍 I had trouble installing Python at first. The instructions said <code>python</code>, but on my system the correct command was <code>python3</code>. That one number made all the difference 🙃.</li>
      <li>🔄 I cloned the wrong GitHub repository into the wrong directory 🤦, which made my project messy. Eventually I recloned the correct one so it was connected properly to GitHub. This taught me to <em>double-check the path</em> before cloning.</li>
      <li>🍺 I discovered that some tools require <code>brew install</code> (on macOS). For example, installing <code>wget</code> is super useful for downloading files directly from the terminal. I want to make a dedicated page for package installs so I don’t forget this in the future 📖.</li>
      <li>🤔 Many hiccups came from not knowing “why” I was doing something. I realized that understanding the reasoning behind each step is just as important as typing the commands.</li>
    </ul>
  </section>

  <section>
    <h2>📖 The Meaning of Commands</h2>
    <ul>
      <li><strong>mkdir</strong> ➝ 📂 make a new directory (like creating a new folder for a project).</li>
      <li><strong>ls</strong> ➝ 📜 list the files in the current directory (like peeking into the folder).</li>
      <li><strong>cd</strong> ➝ 🚪 “change directory” (move into/open a folder to work inside it).</li>
      <li><strong>pwd</strong> ➝ 🧭 print working directory (shows me my exact location in the file tree — like GPS for coding).</li>
      <li><strong>git clone</strong> ➝ 📥 download a repository from GitHub onto my computer.</li>
      <li><strong>brew install</strong> ➝ 🍺 install software packages on macOS, like adding new tools to my toolbox.</li>
    </ul>
  </section>

  <section>
    <h2>📓 Learning Jupyter Notebooks</h2>
    <p>One of the most exciting parts of the day was setting up <strong>Jupyter Notebooks</strong>. I learned that Jupyter is an environment where you can mix:</p>
    <ul>
      <li>➡️ <strong>Code cells</strong> where you can run Python directly and see output immediately 🐍.</li>
      <li>➡️ <strong>Markdown cells</strong> where you can write notes, reflections, or explanations ✍️.</li>
      <li>➡️ <strong>Interactive visuals</strong> like graphs and charts 📊.</li>
    </ul>
    <p>I installed Jupyter using:</p>
    <pre><code>pip install notebook</code></pre>
    <p>Then I launched it with:</p>
    <pre><code>jupyter notebook</code></pre>
    <p>This opened a browser window 🌐 where I could create <code>.ipynb</code> files. I loved how I could document my thought process while testing code in real time. It felt like a coding diary — perfect for learning and experimenting!</p>
  </section>

  <section>
    <h2>💡 Things I Learned</h2>
    <ul>
      <li>📝 <strong>Take notes on everything</strong>. Without notes, it’s easy to forget which commands worked and which didn’t. Jupyter actually helps with this since code and notes live in the same place.</li>
      <li>💾 <strong>Commit often</strong> to GitHub when things are working. That way, if I break something, I can always roll back to a working state instead of starting over.</li>
      <li>🔍 <strong>Understand the “why”</strong>. Don’t just copy-paste commands — know why they’re necessary. For example, using <code>python3</code> instead of <code>python</code> makes sense because my system differentiates between versions.</li>
      <li>🌐 <strong>Tools like brew and wget</strong> make setup smoother. Having a mental map of dependencies avoids a lot of wasted time later.</li>
      <li>📓 <strong>Jupyter is powerful</strong> for beginners because it encourages both experimentation and documentation. It forces me to explain my process while coding.</li>
    </ul>
  </section>
