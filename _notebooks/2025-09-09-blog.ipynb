    <h2>📖 The Meaning of Commands</h2>
    <ul>
      <li><strong>mkdir</strong> ➝ 📂 make a new directory.</li>
      <li><strong>ls</strong> ➝ 📜 list the files in the current directory.</li>
      <li><strong>cd</strong> ➝ 🚪 change directory (move into a folder).</li>
      <li><strong>pwd</strong> ➝ 🧭 print working directory (your location in the file tree).</li>
      <li><strong>git clone</strong> ➝ 📥 download a GitHub repository.</li>
      <li><strong>brew install</strong> ➝ 🍺 install software packages on macOS.</li>
    </ul>
  </section>

  <section>
