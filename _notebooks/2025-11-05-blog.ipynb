    <h2>⚙️ Sprint 2: Fundamentals of JavaScript & Python</h2>
    <p>This sprint was the real turning point for me. I started to connect *computer science theory* to actual code, using ideas from AP CSP’s Big Idea 3. Here’s what I learned concept by concept:</p>

    <ul>
   Variables & Assignments (3.1):</strong> I learned how to store and update values in both Python (<code>x = 5</code>) and JavaScript (<code>let x = 5;</code>). I started using variables not just to hold data, but to make my code flexible and reusable.</li>

</strong> I began grouping related information in arrays/lists. This helped me loop through data — for example, storing player moves in Rock Paper Scissors or columns in Connect 4.</li>

  <strong>Mathematical & Boolean Expressions (3.3–3.5):</strong> These became key to writing game logic — checking if someone won, keeping score, or setting conditions for loops.</li>

  <strong>Conditionals & Nested Conditionals (3.6–3.7):</strong> I used <code>if</code>, <code>else if</code>, and <code>else</code> statements to control the flow of my games. Nested conditionals helped me handle complex logic like “if both players choose the same move, it’s a tie.”</li>

      <li><strong>Iteration (3.8):</strong> Loops became second nature — <code>for</code> loops in JavaScript and <code>while</code> loops in Python. I used iteration to check win conditions and repeat rounds automatically.</li>

      <li><strong>Developing Algorithms (3.9) & Efficiency (3.17):</strong> I learned to break problems into smaller parts — writing functions for specific tasks like checking a win, updating a score, or resetting a board. I also started thinking about how to make those functions faster and avoid repetitive code.</li>

      <li><strong>Procedures (3.12–3.13):</strong> Defining and calling my own functions was the biggest step. I learned to pass parameters, return values, and reuse code cleanly.</li>

      <li><strong>Random Values (3.15):</strong> I used randomness to make Rock Paper Scissors fun — letting the computer choose its move unpredictably each round.</li>
    </ul>

    <p>By the end, I was thinking like a programmer — planning logic before typing code, and using debugging tools and <code>console.log()</code>/<code>print()</code> to trace problems. Each small concept connected into one big picture of how computers think.</p>
