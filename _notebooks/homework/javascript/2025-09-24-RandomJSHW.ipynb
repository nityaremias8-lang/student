```python
%%js
// Hack 3 Coin Flip Streak Tracker

// Clear this cell's output area
element.empty();

// UI
const root = $('<div style="font-family: ui-monospace, SFMono-Regular, Menlo, monospace; line-height:1.4;"></div>').appendTo(element);
root.append('<h3>🪙 Coin Flip Streak Tracker</h3>');
const info = $(`
  <div>
    <button id="flipBtn">🎲 Flip Coin</button>
    <button id="resetBtn" style="margin-left:8px;">🔄 Reset</button>
    <div style="margin-top:12px;">
      <div id="lastFlip">Last flip: —</div>
      <div id="currentStreak">Current streak: 0</div>
      <div id="bestStreak">Best streak: 0</div>
      <div id="flipCount">Total flips: 0</div>
    </div>
    <details style="margin-top:12px;">
      <summary>Show flip history</summary>
      <ul id="history" style="margin-top:8px; padding-left: 20px;"></ul>
    </details>
  </div>
`).appendTo(root);

// State
let currentStreak = 0;
let bestStreak = 0;
let lastResult = null;
let flips = [];

// Helpers
function randCoin() {
  return Math.random() < 0.5 ? 'Heads' : 'Tails';
}

function render() {
  const emojiMap = { Heads: '🪙', Tails: '🎩' };
  $('#currentStreak').text(`Current streak: ${currentStreak}`);
  $('#bestStreak').text(`Best streak: ${bestStreak}`);
  $('#lastFlip').html(`Last flip: ${lastResult ? emojiMap[lastResult] + ' ' + lastResult : '—'}`);
  $('#flipCount').text(`Total flips: ${flips.length}`);

  // Render history as bullet list
  const historyEl = $('#history');
  historyEl.empty();
  flips.forEach(flip => {
    historyEl.append(`<li>${emojiMap[flip]} ${flip}</li>`);
  });
}

// Events
$('#flipBtn').on('click', () => {
  const r = randCoin();
  flips.push(r);
  if (r === lastResult) {
    currentStreak += 1;
  } else {
    currentStreak = 1; // start new streak
  }
  bestStreak = Math.max(bestStreak, currentStreak);
  lastResult = r;
  render();
});

$('#resetBtn').on('click', () => {
  currentStreak = 0;
  bestStreak = 0;
  lastResult = null;
  flips = [];
  render();
});

render();
```

```python
%%js
// Hack 4 — Random Compliment Generator 

// Clear this cell's output area
element.empty();

// UI setup
const root = $('<div style="font-family: Inter, ui-sans-serif, system-ui; line-height:1.5; text-align:center;"></div>').appendTo(element);
root.append('<h3 style="color:#4A4AFC;">💫 Random Compliment Generator</h3>');

const box = $(`
  <div>
    <button id="newCompliment" 
      style="background-color:#4A4AFC; color:white; border:none; border-radius:8px; padding:8px 16px; cursor:pointer; font-size:14px; transition:0.2s;">
      Get a Compliment 💬
    </button>
    <div id="compliment" 
      style="margin-top:14px; padding:14px; border:1px solid #ccc; border-radius:10px; min-height:40px; max-width:400px; margin-left:auto; margin-right:auto; background-color:#f8f9ff;">
    </div>
  </div>
`).appendTo(root);

// Compliment list (fresh set with positive, creative tones)
const compliments = [
  "Your ideas always spark something new.",
  "You bring calm energy to every challenge.",
  "Your curiosity makes learning exciting.",
  "You make others feel seen and heard.",
  "Your resilience is truly impressive.",
  "You have a gift for creative thinking.",
  "You make progress look effortless.",
  "Your optimism lifts everyone around you.",
  "You turn ordinary moments into inspiration.",
  "Your perspective makes a difference.",
  "You have the courage to keep improving.",
  "You think deeply and act with purpose.",
  "Your confidence encourages others to try.",
  "You make growth look graceful.",
  "You’re doing better than you think."
];

// Function to pick a random compliment
function pick() {
  const i = Math.floor(Math.random() * compliments.length);
  return compliments[i];
}

// Animation + display
function showCompliment() {
  const compliment = pick();
  const div = $('#compliment');
  div.fadeOut(150, () => {
    div.text(compliment).fadeIn(250);
  });
}

// Initialize
$('#newCompliment').on('click', showCompliment);
showCompliment();
```
