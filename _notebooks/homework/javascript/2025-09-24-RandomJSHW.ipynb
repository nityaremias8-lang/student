// --- JavaScript Logic ---
let lastFlip = null;
let currentStreak = 0;
let maxStreak = 0;

const compliments = [
  "You're doing amazing! 💪",
  "You're a coding rockstar! ⭐",
  "Your smile is contagious 😄",
  "You're a fast learner! ⚡",
  "Your logic is top-notch 🧠",
  "You bring good energy! 🌟",
  "You’re so creative 🎨",
  "You have a great sense of humor 😂",
  "Your hard work shows 💼",
  "You're unstoppable 🚀",
  "You brighten up the room ☀️",
  "You're a natural problem-solver 🧩"
];

function flipCoin() {
  const flip = Math.random() < 0.5 ? "Heads" : "Tails";
  document.getElementById("coinResult").textContent = `🪙 You flipped: ${flip}`;

  if (flip === lastFlip) {
    currentStreak++;
  } else {
    currentStreak = 1;
  }

  if (currentStreak > maxStreak) {
    maxStreak = currentStreak;
  }

  lastFlip = flip;

  document.getElementById("streakInfo").textContent =
    `🔥 Current Streak: ${currentStreak} ${flip}! | 🏆 Max Streak: ${maxStreak}`;
}

function giveCompliment() {
  const randomIndex = Math.floor(Math.random() * compliments.length);
  const compliment = compliments[randomIndex];
  document.getElementById("complimentBox").textContent = `💬 ${compliment}`;
}

// Bind buttons to functions
document.getElementById("flipBtn").addEventListener("click", flipCoin);
document.getElementById("complimentBtn").addEventListener("click", giveCompliment);
