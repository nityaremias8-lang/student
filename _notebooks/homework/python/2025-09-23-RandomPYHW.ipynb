```python

import random

def dice_rolling_counter():
    rolls = 20
    counts = {i: 0 for i in range(1, 7)}  # Initialize counts for 1 to 6

    print(f"Rolling a 6-sided die {rolls} times...\n")

    for i in range(1, rolls + 1):
        roll = random.randint(1, 6)
        counts[roll] += 1
        print(f"Roll {i}: {roll}")

    print("\nResults:")
    for number, count in counts.items():
        print(f"Number {number}: {count} time(s)")

if __name__ == "__main__":
    dice_rolling_counter()
```

```python
import random

def random_story_generator():
    characters = ["knight", "wizard", "dragon"]
    places = ["castle", "forest", "cave"]
    actions = ["fought", "slept", "ate"]

    print("Generating 5 random stories:\n")
    for i in range(1, 6):
        char = random.choice(characters)
        place = random.choice(places)
        action = random.choice(actions)
        story = f"The {char} went to the {place} and {action}."
        print(f"Story {i}: {story}")

if __name__ == "__main__":
    random_story_generator()
```
