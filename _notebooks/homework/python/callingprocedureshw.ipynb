```python

# 1. Basic Call
def greet():
    print("Hello, world!")

# Call the "greet" procedure here
greet()


# 2. Calling with Parameters
def greet(name):
    print("Hello,", name)

# Put your name and a friend's name into the call!
greet("Alice")
greet("Rachael")


# 3. Call within a Call
def jump():
    print("Jump up!")

def dance():
    # Call the jump procedure multiple times
    print("Spin around!")
    jump()
    jump()
    jump()

dance()


# Let’s try it with numbers!
# Change the numbers below to any two numbers to check the sum
def add_numbers(a, b):
    sum_value = a + b
    return sum_value

answer = add_numbers(5, 7)
print(answer)  # Should print 12


# Now try multiplication
def multiply_numbers(a, b):
    product = a * b
    return product

answer = multiply_numbers(3, 9)
print(answer)  # Should print 27

```
